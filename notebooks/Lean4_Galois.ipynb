**Manuscript alignment:** Lemma 3.1 (Compatibilidad de Galois con proyecciones modulares).

**Epistemic note:** This module certifies the **compatibility** of Galois automorphisms with modular projections, **not** the deterministic linkage of conjugate residues. The latter claim is FALSE in general (see the explicit counterexample in Section 3.2 of the manuscript).

```python
# ==============================================================================
# Module 6: Galois Projection Compatibility
# Manuscript: Lemma 3.1
# Reference Manuscript: Galois Invariants in Cyclotomic Lattice Enumeration:
# A GPU-Accelerated Meet-in-the-Middle Attack on Ring-LWE under Modular Filtration
#
# Author: José Ignacio Peinador Sala (Independent Researcher, Valladolid, Spain)
# ORCID: https://orcid.org/0009-0008-1822-3452
# Reproducibility DOI: https://doi.org/10.5281/zenodo.19920082
# ==============================================================================
import os, subprocess, sys

PROJECT_DIR = "/content/Galois_Invariants_LWE_Lean"
LEAN_FILE = os.path.join(PROJECT_DIR, "Galois_Invariants_LWE_Lean", "AdvancedLatticeSkeletons.lean")
os.makedirs(os.path.dirname(LEAN_FILE), exist_ok=True)

lean_code = """import Mathlib.RingTheory.Ideal.Operations
import Mathlib.Algebra.Ring.Equiv

set_option linter.style.longLine false

/-!
# Module: Galois Projection Compatibility

# PURPOSE
# This module certifies the Galois projection compatibility lemma, which is
# the mathematically correct core of the conjugate residue analysis. It states
# that a Galois automorphism sigma : R -> R preserving an ideal (sigma(p1) = p2)
# is compatible with the canonical projections R -> R/p1 and R -> R/p2.

# IMPORTANT EPISTEMIC NOTE
# This module certifies that the map sigma_bar : R/p1 -> R/p2 induced by sigma is
# WELL-DEFINED. It does NOT certify that knowing pi_p1(s) alone determines
# pi_p2(s) for an arbitrary s in R. The latter claim is FALSE in general.
-/

namespace RingLWE.AdvancedLattice

/--
THEOREM: Galois Projection Compatibility.
Certifies that if sigma : R -> R is a ring automorphism with sigma(p1) = p2,
then for any two elements s1, s2 in R,
    s1 - s2 in p1  ==>  sigma(s1) - sigma(s2) in p2.
Equivalently, the induced map sigma_bar : R/p1 -> R/p2 is well-defined.
-/
theorem galois_projection_compatibility {R : Type*} [CommRing R]
    (p1 p2 : Ideal R) (sigma : R ≃+* R)
    (h_map : p1.map (sigma : R →+* R) = p2)
    (s1 s2 : R) (h_eq : s1 - s2 ∈ p1) :
    sigma s1 - sigma s2 ∈ p2 := by
  rw [← map_sub]
  rw [← h_map]
  exact Ideal.mem_map_of_mem (sigma : R →+* R) h_eq

/--
COROLLARY: Well-definedness of the induced map on quotient rings.
-/
theorem induced_map_well_defined {R : Type*} [CommRing R]
    (p1 p2 : Ideal R) (sigma : R ≃+* R)
    (h_map : p1.map (sigma : R →+* R) = p2) :
    ∀ s1 s2 : R, s1 - s2 ∈ p1 → sigma s1 - sigma s2 ∈ p2 := by
  intro s1 s2 h_eq
  exact galois_projection_compatibility p1 p2 sigma h_map s1 s2 h_eq

end RingLWE.AdvancedLattice
"""

with open(LEAN_FILE, "w") as f: f.write(lean_code)
print(f"Module written to: {LEAN_FILE}\nCompiling...")
env_path = f"/root/.elan/bin:{os.environ['PATH']}"
res = subprocess.run(f"PATH='{env_path}' lake build", shell=True, cwd=PROJECT_DIR, capture_output=True, text=True)

if res.returncode == 0:
    print("Build Successful! 100% Certified (0 sorry assertions).")
    print("Verified Proofs:")
    print("  - galois_projection_compatibility")
    print("  - induced_map_well_defined")
else:
    print(f"Compilation Failed:\n{res.stderr}", file=sys.stderr)
```
